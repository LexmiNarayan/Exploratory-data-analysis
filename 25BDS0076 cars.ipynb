```python
df1=pd.DataFrame({'model':['Ford','Mercedes','Bently'],'mpg':[15.2,20.5,16.4],'cyl':[6,8,4],'disp':[305.0,89.0,180.0],'hp':[130,175,90],'drat':[3.95,4.05,3.65],'wt':[3.455,2.986,3.055],'qsec':[15.78,14.55,16.05],'vs':[0,1,1],'am':[0,1,0],'gear':[3,5,6]})
vertic_concat=pd.concat([df,df1],ignore_index=True)
print("Vertical concatination:")
print(vertic_concat)
```

```python
df2=pd.DataFrame({'year':[2015,2010,2000,2015,2010,2009]})
horiz_concat=pd.concat([df,df2],axis=1)
print("Horizontal Concatination:")
print(horiz_concat)
```

```python
df3=pd.DataFrame({'model':['Mazda RX4','Datsun 710','Valiant','Merc 230'],'year':[2010,2002,2007,2008]})
inner_joint=pd.merge(df,df3,on='model',how='inner')
print(inner_joint)
```

```python
left_joint=pd.merge(df,df3,on='model',how='left')
print(left_joint)
```

```python
right_joint=pd.merge(df,df3,on='model',how='right')
print(right_joint)
```

```python
outer_joint=pd.merge(df,df3,on='model',how='outer')
print(outer_joint)
```

```python
price=pd.DataFrame({'effective':[0.75,0.55,0.68,0.85],'orig_cost':[20000,45000,30000,55000]})
price['market_price']=price['effective']*price['orig_cost']
print(price)
```

```python
s1=pd.Series([15,16,17,18],index=['a','b','c','d'])
s2=pd.Series([8,9,10],index=['f','g','h'])
f=pd.concat([s1,s2],keys=['M1','M2'])
print("\n Heirarchial Series:")
print(f)
print("\nUnstacked Series")
print(f.unstack())
```
